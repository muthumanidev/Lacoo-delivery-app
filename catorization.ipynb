```python
ind=dataset[['Temp_C', 'Dew Point Temp_C', 'Rel Hum_%','Wind Speed_km/h', 'Visibility_km', 'Press_kPa',]]
```

```python
dep=dataset[["Weather"]]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(ind, dep, test_size=0.20, random_state=42)
```

```python
from sklearn import tree
```

```python
clf = tree.DecisionTreeClassifier()
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred=clf.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
cls=classification_report(y_test, y_pred)
```

```python
print(cls)
```

```python
weather_categories = {
    "Clear/Fair": [
        "Mainly Clear", "Mostly Cloudy", "Cloudy", "Clear", "Haze",
        "Snow,Haze"  # since haze can be considered fair/atmospheric, some place in Clear
    ],

    "Fog/Low Visibility": [
        "Fog", "Rain,Fog", "Drizzle,Fog", "Moderate Rain,Fog",
        "Rain Showers,Fog", "Drizzle,Ice Pellets,Fog", "Freezing Fog",
        "Snow Showers,Fog", "Thunderstorms,Rain Showers,Fog",
        "Thunderstorms,Moderate Rain Showers,Fog", "Thunderstorms,Rain,Fog",
        "Freezing Drizzle,Fog", "Freezing Rain,Fog"
    ],

    "Rain": [
        "Rain", "Rain Showers", "Drizzle", "Rain,Haze",
        "Drizzle,Snow" , "Drizzle,Snow,Fog",
        "Rain,Snow,Ice Pellets", "Rain,Ice Pellets", "Rain,Snow,Fog",
        "Rain,Snow Grains", "Rain Showers,Snow Showers"
    ],

    "Snow/Wintry": [
        "Snow", "Snow Showers", "Snow,Fog", "Snow,Blowing Snow",
        "Moderate Snow", "Moderate Snow,Blowing Snow", "Snow,Ice Pellets",
        "Snow Pellets", "Rain,Snow", "Freezing Rain", "Freezing Drizzle",
        "Freezing Drizzle,Snow", "Freezing Rain,Haze", "Freezing Drizzle,Haze",
        "Freezing Rain,Snow Grains"
    ],

    "Thunderstorms": [
        "Thunderstorms", "Thunderstorms,Rain Showers", "Thunderstorms,Rain",
        "Thunderstorms,Heavy Rain Showers"
    ]
}
```

```python
condition_to_category = {}
for cat, conditions in weather_categories.items():
    for cond in conditions:
        condition_to_category[cond] = cat
```

```python
dataset["Category"] = dataset["Weather"].map(condition_to_category)
```
